## ✅ Introduction

### **Emoji Sentiment Analysis with TextBlob**

This notebook analyzes emotional changes across different age groups using emoji-supported text data. Emojis often carry emotional weight and add context to the text. We will use the **TextBlob** library to determine sentiment (positive, negative, or neutral) for each statement.  
The dataset contains short statements representing typical feelings at different ages, and each statement includes an emoji to express the mood.

---

## ✅ Step-by-Step Explanation

### 1: Data Preparation
We define a sample dataset that represents emotions at different age groups. The dataset can be expanded to include more real-world text data, such as social media posts or reviews.

### 2: Sentiment Analysis
We use TextBlob to analyze sentiment polarity.  
Polarity ranges from **-1 (negative)** to **+1 (positive)**.  
Sentiment is classified as:
- **Positive** (polarity > 0)
- **Negative** (polarity < 0)
- **Neutral** (polarity = 0)

### 3: Sentiment Distribution
We calculate the percentage of positive, negative, and neutral sentiments.  
This helps to understand the overall emotional trend across different age groups.

### 4: Visualization
We visualize the sentiment distribution using bar charts.  
The visualization provides a clear understanding of emotional patterns across ages.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from textblob import TextBlob

#  Full text display
pd.set_option('display.max_colwidth', None)
pd.set_option('display.width', 1000)

data = {
    "Age": [10, 20, 25, 25, 30, 40, 50, 60],
    "Text": [
        "At age 10, childhood is happy, playful, and stress free ☺️",
        "At age 20, academic pressure and family stress create mental burden 😐",
        "At 25, failed relationships often cause emotional pain and sadness 💔😢",
        "At 25, focusing on career goals brings motivation and hope 😌",
        "At age 30, work pressure and financial problems cause anxiety and frustration 😔",
        "At age 40, life feels balanced with stability and satisfaction 🙂",
        "At age 50, health issues and exhaustion lead to worry and discomfort 😞",
        "At age 60, peace, gratitude, and happiness become the main priorities 🌸😊"
    ]
}

df = pd.DataFrame(data)

def get_sentiment(text):
    return TextBlob(text).sentiment.polarity

df['Sentiment'] = df['Text'].apply(get_sentiment)

df['Label'] = df['Sentiment'].apply(
    lambda x: 'Positive' if x > 0 else ('Negative' if x < 0 else 'Neutral')
)

display(df)
#=====================================================

print("*"*20)
print("Percentage".center(10))
print("*"*20)

#======================================================

label_counts = df['Label'].value_counts()
percentage = (label_counts / len(df)) * 100
display(percentage)

#=======================================================
                     # Bar Plot
#=======================================================

print(" ")
print("*"*50)
print("Bar plot".center(50))
print("*"*50)
print(" ")

#========================================================

percentage.plot(kind='bar')
plt.title("Emotional Distribution Across Age Groups")
plt.xlabel("Emotion Category")
plt.ylabel("Percentage (%)")
plt.show()
```

----
## ✅ Conclusion 

The sentiment distribution shows that 50% of the statements are neutral, while 25% are positive and 25% are negative.
This indicates a balanced emotional representation, with neutral expressions being the most common across different age groups. The dataset effectively captures mixed emotional states, reflecting real-life variations in feelings.
